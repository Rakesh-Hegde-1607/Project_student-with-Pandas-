#### Section 0: Data Loading

```python
import pandas as pd
df = pd.read_excel("aiml_attendance.xlsx")
df
```

#### Section 1: Data Inspection

```python
"""Q1. Display the first 10 rows, the shape of the DataFrame, and the data types of each column."""
#-----------------------------Write Solution Code-----------------------------
import pandas as pd

print(df.head(10))
print(df.shape)
print(df.dtypes)
```

```python
"""Q2. Check for missing values in each column. Identify which columns have null values (especially Last Present Date with "Never") and report the count. """
#-----------------------------Write Solution Code-----------------------------
print(df.isnull().sum())
print((df == "Never").sum())
print(df['Last Present Date'].value_counts(dropna=False))
```

```python
"""Q3. Rename the columns to snake_case (e.g., Roll No → roll_no, Student Name → student_name, Attendance % → attendance_pct). Display the updated column list."""
#-----------------------------Write Solution Code-----------------------------
df.columns = df.columns.str.strip().str.lower().str.replace('%','pct').str.replace(r'[^\w]+','_', regex=True).str.strip('_')
print(df.columns.tolist())
```

```python
"""Q4. Convert Attendance % from string format (e.g., "64%") to a numeric float column (0.64 or 64.0). Also convert Last Present Date to datetime, handling "Never" as NaT."""
#-----------------------------Write Solution Code-----------------------------
df['attendance_pct'] = df['attendance_pct'].astype(str).str.replace('%','', regex=False).str.strip()
df['attendance_pct'] = pd.to_numeric(df['attendance_pct'], errors='coerce')

df['last_present_date'] = df['last_present_date'].replace('Never', pd.NaT)
df['last_present_date'] = pd.to_datetime(df['last_present_date'], errors='coerce')

print(df[['attendance_pct', 'last_present_date']].head(10))
```

#### Section 2: Filtering & Selection

```python
"""Q5. Filter and display all students whose Status is "Critical". Sort them by Attendance % in ascending order."""
#-----------------------------Write Solution Code-----------------------------
critical_df = df[df['status'] == 'Critical'].sort_values('attendance_pct')
print(critical_df)
```

```python
"""Q6. Extract all students who have 0 present days (fully absent). List their roll numbers, names, and consecutive absences."""
#-----------------------------Write Solution Code-----------------------------
absent_df = df[df['present'] == 0][['roll_no','student_name','consecutive_absences']]
print(absent_df)
```

```python
"""Q7. Using .loc and .iloc, display:
(a) All "Safe" students using .loc
(b) The first 15 rows and columns 2-6 using .iloc """
#-----------------------------Write Solution Code-----------------------------
print(df.loc[df['status'] == 'Safe'])
print(df.iloc[0:15, 2:6])
```

```python
"""Q8. Find all students whose Consecutive Absences is greater than or equal to 5. Display their name, status, and last present date."""
#-----------------------------Write Solution Code-----------------------------
print(df[df['consecutive_absences'] >= 5][['student_name','status','last_present_date']])
```

### Section 3: Grouping & Aggregation 

```python
"""Q9. Group the data by Status and compute: 
a) Count of students 
b) Average attendance %  
c) Maximum consecutive absences"""
#-----------------------------Write Solution Code-----------------------------
print(df.groupby('status').agg(
    student_count=('roll_no','count'),
    avg_attendance=('attendance_pct','mean'),
    max_consecutive_absences=('consecutive_absences','max')
))
```

```python
"""Q10. Create a new column Attendance Category:
≥ 90% → "Excellent"
75-89% → "Good"
60-74% → "Average"
< 60% → "Poor"
Then use groupby() to count students in each category."""
#-----------------------------Write Solution Code-----------------------------
def get_category(x):
    if x >= 90:
        return "Excellent"
    elif x >= 75:
        return "Good"
    elif x >= 60:
        return "Average"
    else:
        return "Poor"
    df['attendance_category'] = df['attendance_pct'].apply(get_category)
    print(df.groupby('attendance_category').size())
```

```python
"""Q9. Group the data by Status and compute: 
a) Count of students 
b) Average attendance %  
c) Maximum consecutive absences"""
#-----------------------------Write Solution Code-----------------------------
print(df.groupby('status').agg(
    student_count=('roll_no','count'),
    avg_attendance=('attendance_pct','mean'),
    max_consecutive_absences=('consecutive_absences','max')
))
```

```python
"""Q10. Create a new column Attendance Category:
≥ 90% → "Excellent"
75-89% → "Good"
60-74% → "Average"
< 60% → "Poor"
Then use groupby() to count students in each category."""
#-----------------------------Write Solution Code-----------------------------
def get_category(x):
    if x >= 90:
        return "Excellent"
    elif x >= 75:
        return "Good"
    elif x >= 60:
        return "Average"
    else:
        return "Poor"

df['attendance_category'] = df['attendance_pct'].apply(get_category)
print(df.groupby('attendance_category').size())
```

```python
"""Q11. Calculate the total number of classes conducted (sum of Total Classes) and the total present/absent counts across all students using .agg()."""
#-----------------------------Write Solution Code-----------------------------
print(df.agg(
    total_classes=('total_classes','sum'),
    total_present=('present','sum'),
    total_absent=('absent','sum')
))
```

```python
"""Q12. Group by Status and use .agg() to find:
a)min, max, and mean of Present
b)sum of Absent"""
#-----------------------------Write Solution Code-----------------------------
print(df.groupby('status').agg(
    min_present=('present','min'),
    max_present=('present','max'),
    mean_present=('present','mean'),
    sum_absent=('absent','sum')
))
```

### Section 4: Feature Engineering & Transformation

```python
"""Q13. Create a new column Risk Score using the formula:
Risk Score = (Absent * 0.4) + (Consecutive Absences * 0.6)
Display the top 10 students with the highest risk score."""
#-----------------------------Write Solution Code-----------------------------
df['risk_score'] = (df['absent'] * 0.4) + (df['consecutive_absences'] * 0.6)
print(df.sort_values('risk_score', ascending=False).head(10))
```

```python
"""Q14. Add a column Classes Needed (Numeric) by extracting the numeric value from the Classes Needed (75%) column (e.g., "5 more" → 5, "✓ Met" → 0)."""
#-----------------------------Write Solution Code-----------------------------
print(df.columns.tolist())

target_col = [c for c in df.columns if 'classes' in c and 'needed' in c][0]
print(target_col)

df['classes_needed_numeric'] = df[target_col].astype(str).str.extract(r'(\d+)').fillna(0).astype(int)

print(df[['{}'.format(target_col), 'classes_needed_numeric']].head(10))
```

```python
"""Q15. Using apply() with a lambda function, create a column Attendance Flag that returns:
a)"Defaulter" if attendance < 75%
b)"Safe" otherwise """
#-----------------------------Write Solution Code-----------------------------
df['attendance_flag'] = df['attendance_pct'].apply(lambda x: "Defaulter" if x < 75 else "Safe")
print(df['attendance_flag'])
```

```python
"""Q16. Use pd.cut() to bin students into attendance ranges: [0-50, 50-75, 75-90, 90-100] and label them as ["Critical", "Warning", "Good", "Excellent"]."""
#-----------------------------Write Solution Code-----------------------------
df['attendance_bin'] = pd.cut(df['attendance_pct'], bins=[0,50,75,90,100], labels=["Critical","Warning","Good","Excellent"])
print(df['attendance_bin'])
```

### Section 5: Advanced Analysis 

```python
"""Q17. Using pivot_table(), create a summary showing the count of students by Status and Attendance Category (created in Q10)."""
#-----------------------------Write Solution Code-----------------------------
print(pd.pivot_table(df, index='status', columns='attendance_category', values='roll_no', aggfunc='count', fill_value=0))
```

```python
"""Q18. Identify duplicate student names (e.g., "ANANYA", "DISHA") and display all rows where the base name repeats. Use string manipulation to extract the base name."""
#-----------------------------Write Solution Code-----------------------------
df['base_name'] = df['student_name'].astype(str).str.split().str[0]
dup_names = df['base_name'][df['base_name'].duplicated(keep=False)]
print(df[df['base_name'].isin(dup_names)].sort_values('base_name'))
```

```python
""""Q19. Merge this dataset with a hypothetical Student_Details DataFrame containing Roll No, Branch, and Semester. Perform an inner join and display the merged result."""
#-----------------------------Write Solution Code-----------------------------
student_details = pd.DataFrame({
    'roll_no': df['roll_no'].unique(),
    'branch': 'AIML',
    'semester': 4
})
merged_df = pd.merge(df, student_details, on='roll_no', how='inner')
print(merged_df.head())
```

```python
"""Q20. Export the following to separate CSV files:
(a) All "Critical" students
(b) Summary statistics grouped by Status
(c) Top 10 students by Risk Score"""
#-----------------------------Write Solution Code-----------------------------

print(df.columns.tolist())

status_col = [c for c in df.columns if 'status' in c][0]
roll_col = [c for c in df.columns if 'roll' in c][0]

critical_students = df[df[status_col] == 'Critical']
critical_students.to_csv('critical_students.csv', index=False)

summary = df.groupby(status_col).agg(
    avg_attendance=('attendance_pct','mean'),
    total_students=(roll_col,'count')
)
summary.to_csv('summary_by_status.csv')

top10_risk = df.sort_values('risk_score', ascending=False).head(10)
top10_risk.to_csv('top10_risk_score.csv', index=False)

print("Files exported successfully")
print(critical_students.shape)
print(summary)
print(top10_risk.shape)
```
